## Question 1
Write a function that returns strings with length greater than K.


```python
def long_strings(words, k):
    result = []
    for word in words:
        if len(word) > k:
            result.append(word)
    return result

words = ["apple", "pie", "banana"]
k = 3
print(long_strings(words, k))
```

## Question 2
Write a function to calculate the area of a circle or rectangle.


```python
def area(shape, *dimensions):
    if shape == "circle":
        radius = dimensions[0]
        return 3.14 * radius * radius
    elif shape == "rectangle":
        length = dimensions[0]
        width = dimensions[1]
        return length * width
    else:
        return "Invalid shape"

print("Circle Area:", area("circle", 5))
print("Rectangle Area:", area("rectangle", 4, 6))
```

## Question 5
Find the sum of numbers in a list and ignore strings and booleans.


```python
items = [10, "apple", 20, True, 5]
total = 0

for item in items:
    if isinstance(item, (int, float)) and not isinstance(item, bool):
        total += item

print("Total Sum:", total)
```

## Question 6
Create a Restaurant class and add menu items, table reservations, and customer orders.


```python
class Restaurant:
    def __init__(self):
        self.menu_items = {}
        self.book_table = []
        self.customer_orders = []

    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price

    def book_tables(self, customer, table_number):
        self.book_table.append({"customer": customer, "table": table_number})

    def customer_order(self, customer, item):
        if item in self.menu_items:
            self.customer_orders.append({"customer": customer, "item": item})
        else:
            print(item, "is not on the menu")

    def print_menu(self):
        print("Menu:", self.menu_items)

    def print_reservations(self):
        print("Table Reservations:", self.book_table)

    def print_orders(self):
        print("Customer Orders:", self.customer_orders)


restaurant = Restaurant()

restaurant.add_item_to_menu("Burger", 120)
restaurant.add_item_to_menu("Pizza", 180)
restaurant.add_item_to_menu("Juice", 35)

restaurant.book_tables("Ahmed", 1)
restaurant.book_tables("Mona", 2)

restaurant.customer_order("Ahmed", "Burger")
restaurant.customer_order("Mona", "Juice")

restaurant.print_menu()
restaurant.print_reservations()
restaurant.print_orders()
```

## Question 7
Create Customer, Bike, and MainBikeRental classes.


```python
class Customer:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_name(self):
        return self.name

    def set_name(self, name):
        self.name = name

    def get_age(self):
        return self.age

    def set_age(self, age):
        self.age = age


class Bike:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_name(self):
        return self.name

    def set_name(self, name):
        self.name = name

    def get_price(self):
        return self.price

    def set_price(self, price):
        self.price = price


class MainBikeRental:
    def __init__(self, bikes=None):
        self.bikes = bikes if bikes is not None else []
        self.rentals = []

    def get_bikes(self):
        return self.bikes

    def set_bikes(self, bikes):
        self.bikes = bikes

    def get_rentals(self):
        return self.rentals

    def set_rentals(self, rentals):
        self.rentals = rentals

    def addBike(self, bike, quantity=1):
        self.bikes.append({"bike": bike, "quantity": quantity})

    def totalCost(self):
        total = 0
        for item in self.bikes:
            total += item["bike"].get_price() * item["quantity"]
        return total

    def requestBike(self, customer, number_of_bikes):
        if customer.get_age() <= 6:
            print("Customer must be older than 6 years.")
            return

        available = sum(item["quantity"] for item in self.bikes)
        if number_of_bikes <= available:
            remaining = number_of_bikes
            for item in self.bikes:
                take = min(item["quantity"], remaining)
                item["quantity"] -= take
                remaining -= take
                if remaining == 0:
                    break
            self.rentals.append({"customer": customer, "number": number_of_bikes})
            print("Bikes rented successfully.")
        else:
            print("Please wait 10 mins.")

    def returnBike(self, rentalTime, num_of_bikes):
        bill = rentalTime * 40
        self.rentals.append({"returned_bikes": num_of_bikes, "bill": bill})
        if self.bikes:
            self.bikes[0]["quantity"] += num_of_bikes
        else:
            self.bikes.append({"bike": Bike("Bike", 0), "quantity": num_of_bikes})
        print("Total bill:", bill)
        return bill


customer = Customer("Ahmed", 20)
bike = Bike("Mountain Bike", 500)

shop = MainBikeRental()
shop.addBike(bike, 5)

print("Total cost of bikes in store:", shop.totalCost())
shop.requestBike(customer, 2)
shop.returnBike(3, 2)
```
